```python
import random
```

```python
def step(p):
  return 1 if random.random() < p else -1
```

```python
# print(step(0.66))
```

```python
for _ in range(100):
  print(step(0.3))
```

```python
results2 = [step(0.3) for _ in range(100)]
print(sum(results2) / len(results2))
```

```python
print(results2.count(1) / len(results2))
```

```python
print(results2.count(-1) / len(results2))
```

```python
def simulate(n0, N, p):
  position = n0
  steps = 0
  while position != 0 and position != N:
    position += step(p)
    steps += 1
  return steps
```

```python
print(simulate(5, 10, 0.5))
```
