```python
import random
import matplotlib.pyplot as plt
import statistics
```

```python
def step(p):
  return 1 if random.random() < p else -1
```

```python
p = 0.3
results = [step(p) for _ in range(1000)]
print("+1:", results.count(1), " −1:", results.count(-1))
```

```python
def simulate(n0, N, p):
  position = n0
  steps = 0
  while position != 0 and position != N:
    position += step(p)
    steps += 1
  return steps
```

```python
print(simulate(5, 10, 0.5))
```

```python
results2 = [simulate(5, 10, 0.3) for _ in range(100)]
print(statistics.mean(results2))
```
