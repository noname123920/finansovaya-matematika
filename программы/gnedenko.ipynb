```python
import random
import statistics
```

```python
def step(p):
  return 1 if random.random() < p else -1
```

```python
def simulate(n0, N, p):
  position = n0
  while position != 0 and position != N:
    position += step(p)
  return position
```

# Честная игра:

```python
p = 0.5
n0 = 5
N = 10
trials = 10000
```

```python
outcomes = [simulate(n0, N, p) for _ in range(trials)]
prob_ruin = outcomes.count(0) / trials
```
