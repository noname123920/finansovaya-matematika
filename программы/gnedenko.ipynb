```python
import random
import statistics
```

```python
def step(p):
  return 1 if random.random() < p else -1
```

```python
def simulate(n0, N, p):
  position = n0 # следим за игроком (деньгами) игрока A
  while position != 0 and position != N:
    position += step(p)
  return position
```

# Честная игра:

```python
p = 0.5
n0 = 5
N = 10
trials = 10000

outcomes = [simulate(n0, N, p) for _ in range(trials)]
prob_ruin_1 = outcomes.count(0) / trials
print(prob_ruin_1)
```

```python
a = n0 # начальный капитал игрока A
b = N - n0 # начальный капитал игрока B
print(b / (a + b)) # вероятность разорения игрока A
```

# Нечестная игра:

```python
p = 0.6
q = 1 - p
outcomes = [simulate(n0, N, p) for _ in range(trials)]
prob_ruin_2 = outcomes.count(0) / trials
print(prob_ruin_2)
```

```python
print((1 - (p/q)**b) / (1 - (p/q)**(a+b))) # вероятность разорения игрока A
```
